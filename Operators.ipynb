Arithmetic,Assignment,Comparison,Logical,Identity,Membership all are types of operators.

```python
#Arithmetic Operators
n_1 = 20
n_2 = 10

#add
sum = n_1+n_2
print(sum)

#sub
sub = n_1-n_2
print(sub)

#multiply
prod = n_1*n_2
print(prod)

#divide
div = n_1/n_2
print(div)

#exponent
exp = n_1**n_2
print(exp)

# modulus(remainder)
mod = n_1%n_2
print(mod)



```

Assignment Operators

```python
a=5
print(a)

a+=5
print(a)
```

```python
b=5
print(b)
b-=2
print(b)
```

+=,  -=,  *=,  **=,  /=,  %=

Comparison Operators

```python
a=5
b=10

print(a==b) #equal to
print(a!=b) #not equal to
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

Logical Operators - AND,OR,NOT

```python
a=10
print(a>10 and a>5)

print(a>10 or a>5)

print(not(a>8 and a>5))
```

Identity Operators  - is,is not

```python
x=5
y=5
print(x is y)
print(x is not y)

z=10
print(x is z)
print(x is not z)
```

Membership Operators - in,not in

```python
a=5
b=10
c=[1,22,3,4,5,6,7,8,9]

print(a in c)
print(a not in c)
print(b in c)
print(b not in c)
```
